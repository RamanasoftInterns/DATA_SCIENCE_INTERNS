# Breast Cancer Wisconsin Diagnostic Dataset

## Dataset Overview

The Breast Cancer Wisconsin Diagnostic Dataset is a healthcare and machine learning dataset used for binary classification problems. The dataset helps classify breast tumors as either malignant (cancerous) or benign (non-cancerous) based on numerical features extracted from digitized images of breast cell nuclei.

---

# Dataset Information

| Property | Value |
|---|---|
| Number of Rows | 569 |
| Number of Columns | 32 |
| Numerical Columns | 30 |
| Categorical Columns | 1 |
| Identifier Columns | 1 |
| Missing Values | No Missing Values |
| Dataset Type | Structured Tabular Dataset |
| Problem Type | Binary Classification |

---

# Target Variable

| Column Name | Description |
|---|---|
| diagnosis | Tumor diagnosis |

Possible values:

| Value | Meaning |
|---|---|
| M | Malignant (Cancerous) |
| B | Benign (Non-Cancerous) |

# Feature Categories

The 30 numerical features are grouped into three categories:

| Suffix | Meaning |
|---|---|
| _mean | Mean value |
| _se | Standard Error |
| _worst | Worst/Largest value |

---

# Feature Descriptions

## Radius Features

Radius measures the distance from the center to the edge of the nucleus.

- radius_mean
- radius_se
- radius_worst

---

## Texture Features

Texture measures variation in gray-scale intensity.

- texture_mean
- texture_se
- texture_worst

---

## Perimeter Features

Perimeter represents the boundary length of the nucleus.

- perimeter_mean
- perimeter_se
- perimeter_worst

---

## Area Features

Area represents the size of the nucleus.

- area_mean
- area_se
- area_worst

---

## Smoothness Features

Smoothness measures local variation in radius lengths.

- smoothness_mean
- smoothness_se
- smoothness_worst

---

## Compactness Features

Compactness measures the compactness of the cell structure.

- compactness_mean
- compactness_se
- compactness_worst

---

## Concavity Features

Concavity measures the severity of concave portions of the contour.

- concavity_mean
- concavity_se
- concavity_worst

---

## Concave Points Features

Concave points represent the number of concave portions in the contour.

- concave_points_mean
- concave_points_se
- concave_points_worst

---

## Symmetry Features

Symmetry measures the symmetry of the cell nucleus.

- symmetry_mean
- symmetry_se
- symmetry_worst

---

## Fractal Dimension Features

Fractal dimension measures boundary complexity.

- fractal_dimension_mean
- fractal_dimension_se
- fractal_dimension_worst
